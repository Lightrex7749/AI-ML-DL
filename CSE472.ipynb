```python
# Import WordNet Lemmatizer and Demonstrate Use
import nltk
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')
lemmatizer=WordNetLemmatizer()
words=['eating','running','eats','ate','adjustable','rafting','ability','meeting']

for word in words:
  print(word,"|",lemmatizer.lemmatize(word,pos='v'))
  print(word,"|",lemmatizer.lemmatize(word,pos='a'))
  print(word,"|",lemmatizer.lemmatize(word,pos='n'))
```

```python
# Compare Stemming and Lemmatization
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
lst='running painting walking dressing likely children whom good ate fishing'.split()
stemmer=PorterStemmer()
lemmatizer=WordNetLemmatizer()
for word in lst:
  print(word,"|",stemmer.stem(word),"   ==   ",lemmatizer.lemmatize(word,pos="v"))
  print(word,"|",stemmer.stem(word),"   ==   ",lemmatizer.lemmatize(word,pos="a"))
  print(word,"|",stemmer.stem(word),"   ==   ",lemmatizer.lemmatize(word,pos="n"))
```

```python
# Apply Stemming and Lemmatization to a Text
# Convert the given given test into its base form using both stemming and lemmatization.text="""Latha is very multi talented girl.She is good at many skills like dancing, running, singing, playing.She also likes eating Pav Bhagi. she has a habit of fishing and swimming too.Besides all this, she is a wonderful at cooking too."""
import nltk
from nltk.stem import PorterStemmer,WordNetLemmatizer
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')
nltk.download('wordnet')
lemma=WordNetLemmatizer()
stemm=PorterStemmer()
text="Latha is very multi talented girl.She is good at many skills like dancing, running, singing, playing.She also likes eating Pav Bhagi. she has a habit of fishing and swimming too.Besides all this, she is a wonderful at cooking too."
words=word_tokenize(text)
for word in words:
  print(word,"->",lemma.lemmatize(word))
for word in words:
  print(word,"->",stemm.stem(word))
```

```python
text ="the qiuck fox humps over the lazy dog"
words=text.split()
print("words:",words)
```

```python
# a nearual network only understand numbers not text
# so we give each unique word an ID

vocab =sorted(set(words))
vocab_size=len(words)

word_to_id={}
id_to_word={}
for i ,w in enumerate(vocab):
  word_to_id[w]=i
  id_to_word[i]=w

print("vocabalary: ",words)
print("vocabalary size:",vocab_size)
print("word to id:",word_to_id)
print("id to word:",id_to_word)
```

```python
import numpy as np
def one_hot(words):
  vector=np.zeros((vocab_size))
  vector[word_to_id[words]]=1
  return vector

print("ONe hot vector for fox: ")
print(one_hot("fox"))
```

```python
window = 1
cbow_examples = []
for i in range(len(words)): # 'i' is the index of the target word
  target = words[i]
  context = []
  for j in range(-window, window + 1): # 'j' is the offset from the target
    if j == 0: # Skip the target word itself
      continue

    neighbor_index = i + j
    if 0 <= neighbor_index < len(words): # Check if the neighbor index is within bounds
      context.append(words[neighbor_index]) # Append the actual word, not the context list
  cbow_examples.append((context, target)) # Store the context-target pair

print(f"Number of CBOW training examples: {len(cbow_examples)}")
print()
print("first 3 examples context -> target")
for context, target in cbow_examples[:3]:
  print(f"{context} -> {target}")
```
