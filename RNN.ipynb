```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Flatten, Input, SimpleRNN, LSTM
```

```python
raw_seq=np.array(range(1,13))
```

```python
from typing_extensions import Sequence
def make_dataset(sequence,window_size=3):
  X,y=[],[]
  for i in range(len(sequence)-window_size):
    X.append(sequence[i:i+window_size])
    y.append(sequence[i+window_size])
  return np.array(X),np.array(y)

window_size=3
X,y=make_dataset(raw_seq,window_size)
X= X/30.0
y= y/30.0

X=X.reshape((X.shape[0],X.shape[1],1))
print("sample",X[0].flatten())
print("output:",y[0])
```

```python
model=Sequential()
model.add(SimpleRNN(units=32,activation="tanh",input_shape=(window_size,1)))
model.add(Dense(1))
model.compile(optimizer="adam",loss="mse")
model.summary()

model.fit(X,y,epochs=2000,verbose=0)
print("training completed")

test_input=np.array([27,28,29])/30.0
test_input=test_input.reshape((1,window_size,1))
pred=model.predict(test_input)
print(pred)

predicted=model.predict(test_input,verbose=0)
print(f"\nGivern [27,28,29] predict next number: {predicted[0][0]*30:.2f}")
print("(expected:30)")
```
